## 4. Filtering with conditions

A comparison on a column returns a boolean Series. Passing it back into the DataFrame keeps only rows where the condition is True. This is the same boolean masking idea from Day 1, now on a table.

Combine conditions with `&` (and) and `|` (or). Wrap each condition in parentheses.

```python
# health_beauty products that sold more than 15 units in a month
print("Shape", retail.shape)
mask = (retail["product_category_name"] == "health_beauty") | (retail["qty"] > 15)
strong_month = retail[mask]
print("Rows found:", len(strong_month))
strong_month[["product_id", "month_year", "unit_price", "qty"]].head()
```

## 5. Summaries and grouping

`.value_counts()` counts how often each category appears.
`.groupby()` splits the table by a category, then applies a summary to each group. This split, apply, combine pattern is one of the most useful tools in pandas.

```python
print(retail["product_category_name"].unique())
```

```python
print(retail["product_category_name"].value_counts())
```

```python
# Mean unit price per category
average_price_product_category = retail.groupby("product_category_name")["unit_price"].mean().round(2).sort_values(ascending=False)
average_price_product_category
```

## 6. Checking and handling missing data

Real data can have holes, and pandas marks them as `NaN`. The first step is always to check, whatever the source. This dataset happens to be a curated export with no gaps, which is worth confirming rather than assuming.

```python
retail.info()
```

```python
# Count missing values per column
retail.isna().sum().sum()
```

Zero missing values across all 30 columns. That is not always the case, so the habit still matters: `.isna().sum()` per column, then decide. Since this dataset came already aggregated to the product-month level by whoever built it, most of the cleaning work happened before it reached us, but you should never take that on faith.

To keep the two repair strategies fresh, here is how you would apply them if a column did have gaps, using `freight_price` as the example column:

```python
# Option A: drop rows with a missing value in a given column
would_drop = retail.dropna(subset=["freight_price"])
print("Rows if we dropped missing freight_price:", would_drop.shape[0], "(unchanged here, no gaps exist)")

# Option B: fill missing values in a column with its median
filled_example = retail.copy()
filled_example["freight_price"] = filled_example["freight_price"].fillna(filled_example["freight_price"].mean())
print("Missing freight_price after fill:", filled_example["freight_price"].isna().sum())
```

Dropping is simple but loses rows; filling keeps rows but invents values. The right choice depends on how much is missing and why. We carry `retail` forward unchanged for the rest of the week, since it is already complete. Generally, a 5% threshold rule is followed for handling missing data:
1. Less than 5% missing: We can safely drop the rows with missing values, as this small loss will not significantly impact the statistical power of the dataset.
2. More than 5% missing: We avoid dropping data to prevent losing valuable information. Instead, we fill (impute) the missing values based on the column's data type:
    Numerical columns: Filled using the mean (for normally distributed data) or median (if skewed or containing outliers).<br>
    Categorical columns: Filled using the mode (the most frequent value).

## Recap

- A Series is one column, a DataFrame is a table of aligned Series.
- Inspect first: `shape`, `head`, `info`, `describe`.
- Select with `.loc` (labels) and `.iloc` (positions); filter with boolean conditions joined by `&` and `|`.
- `groupby` and `value_counts` summarise categories.
- Always count missing values before assuming there are none; decide per column whether to drop or fill.
- `read_csv` and `to_csv` move data in and out; use `index=False` when writing.
